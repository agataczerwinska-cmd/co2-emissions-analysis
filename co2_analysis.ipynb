# Exploratory analysis of global CO2 emissions using World Bank ESG data

```
import pandas as pd
esg = pd.read_csv("ESGData.csv")
esg.head()
```

```
esg[esg["Indicator Name"].str.contains("CO2 emissions", case=False, na=False)]
```

```
co2 = esg[esg["Indicator Name"] == "CO2 emissions (metric tons per capita)"]
co2.head()
```

```
co2_countries = co2[~co2["Country Name"].isin([
    "Arab World", "Caribbean small states", "Central Europe and the Baltics",
    "Early-demographic dividend", "East Asia & Pacific", "East Asia & Pacific (excluding high income)",
    "East Asia & Pacific (IDA & IBRD)", "Euro area", "Europe & Central Asia",
    "Europe & Central Asia (excluding high income)", "Europe & Central Asia (IDA & IBRD)",
    "European Union", "Fragile and conflict affected situations", "Heavily indebted poor countries (HIPC)",
    "High income", "Low & middle income", "Low income", "Lower middle income",
    "Middle income", "OECD members", "South Asia", "Sub-Saharan Africa",
    "Sub-Saharan Africa (excluding high income)", "Upper middle income", "World"
])]
```

```
co2_countries = co2_countries[co2_countries["Country Name"] != "North America"]
```

```
co2_countries[["Country Name", "2014"]].sort_values("2014", ascending=False).head(15)
```

```
import matplotlib.pyplot as plt

top15 = co2_countries[["Country Name", "2014"]].sort_values("2014", ascending=False).head(15)

plt.figure(figsize=(10,6))
plt.barh(top15["Country Name"], top15["2014"])
plt.xlabel("CO2 emissions (metric tons per capita)")
plt.title("Top 15 CO2 emitters per capita, 2014")
plt.gca().invert_yaxis()  # so #1 is at the top
plt.show()
```

```
chile = co2_countries[co2_countries["Country Name"] == "Chile"]
chile
```

```
year_columns = [str(y) for y in range(1990, 2015)]  # pick a reasonable range
chile_long = chile.melt(id_vars=["Country Name"], value_vars=year_columns,
                          var_name="Year", value_name="CO2_per_capita")
chile_long
```

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='CO2_per_capita', data=chile_long)
plt.title('CO2 Emissions per Capita in Chile (1990-2014)')
plt.xlabel('Year')
plt.ylabel('CO2 Emissions (metric tons per capita)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
